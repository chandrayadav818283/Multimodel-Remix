```python
!pip install folium osmnx

import osmnx as ox
import folium

# Define the origin and destination
origin_place = "Denton, TX"
destination_address = "3422 Dryden Road, Fort Worth, TX"

# Get the graph for the area
G = ox.graph_from_address(destination_address, dist=100000, dist_type='network', network_type='drive')

# Geocode the origin and destination to get their coordinates
orig_latlng = ox.geocode(origin_place)
dest_latlng = ox.geocode(destination_address)

# Find the nearest nodes in the graph to the origin and destination
orig_node = ox.distance.nearest_nodes(G, orig_latlng[1], orig_latlng[0])
dest_node = ox.distance.nearest_nodes(G, dest_latlng[1], dest_latlng[0])

# Calculate the shortest path
route = ox.shortest_path(G, orig_node, dest_node, weight='length')

# Create a Folium map centered at the approximate midpoint of the route
# This makes sure both origin and destination are visible initially
route_nodes_coords = [(G.nodes[node]['y'], G.nodes[node]['x']) for node in route]
if route_nodes_coords:
    avg_lat = sum(p[0] for p in route_nodes_coords) / len(route_nodes_coords)
    avg_lon = sum(p[1] for p in route_nodes_coords) / len(route_nodes_coords)
    m = folium.Map(location=[avg_lat, avg_lon], zoom_start=10)
else:
    m = folium.Map(location=[orig_latlng[0], orig_latlng[1]], zoom_start=10)

# Add the route to the map
m = ox.plot_route_folium(G, route, m)

# Add markers for origin and destination
folium.Marker(
    location=[orig_latlng[0], orig_latlng[1]],
    popup=origin_place,
    icon=folium.Icon(color='green')
).add_to(m)

folium.Marker(
    location=[dest_latlng[0], dest_latlng[1]],
    popup=destination_address,
    icon=folium.Icon(color='red')
).add_to(m)

# Display the map
m
```

```python
route_length_meters = ox.utils_graph.get_route_edge_attributes(G, route, 'length')
route_length_miles = sum(route_length_meters) * 0.000621371 # Convert meters to miles

print(f"Total route length: {route_length_miles:.2f} miles")
```
