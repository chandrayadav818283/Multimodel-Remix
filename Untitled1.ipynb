## Variational Quantum Classifier (VQC) Example with Qiskit

This example demonstrates a basic VQC setup for a simple binary classification problem.

```python
# Install qiskit and qiskit-machine-learning if not already installed
!pip install qiskit qiskit-machine-learning matplotlib

# Import necessary libraries
import numpy as np
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split

from qiskit.algorithms.optimizers import COBYLA
from qiskit.circuit.library import ZZFeatureMap, RealAmplitudes
from qiskit_machine_learning.algorithms import VQC
from qiskit.utils import QuantumInstance, algorithm_globals
from qiskit import BasicAer

import matplotlib.pyplot as plt
```

### 1. Generate Synthetic Dataset

We'll use the `make_moons` dataset from scikit-learn for a simple non-linear classification problem.

```python
# Generate a moon-shaped dataset
X, y = make_moons(n_samples=100, noise=0.1, random_state=42)

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Visualize the dataset
plt.scatter(X[y == 0, 0], X[y == 0, 1], label='Class 0', marker='o')
plt.scatter(X[y == 1, 0], X[y == 1, 1], label='Class 1', marker='x')
plt.title('Synthetic Moon-shaped Dataset')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

print(f"Training data shape: {X_train.shape}, {y_train.shape}")
print(f"Testing data shape: {X_test.shape}, {y_test.shape}")
```

### 2. Define Feature Map and Ansatz

*   **Feature Map (`ZZFeatureMap`):** Encodes the classical data into quantum states. The number of qubits is determined by the number of features in our dataset (2 in this case).
*   **Ansatz (`RealAmplitudes`):** A parameterized quantum circuit that acts as the learning model. Its parameters will be optimized during training. We'll use 2 qubits and 1 repetition (layer).

```python
num_features = X_train.shape[1] # Number of features (dimensions) in our data

# Define the Feature Map (data encoding circuit)
feature_map = ZZFeatureMap(feature_dimension=num_features, reps=1, entanglement='linear')
print("Feature Map Circuit:")
display(feature_map.draw('mpl'))

# Define the Ansatz (variational form / learning circuit)
ansatz = RealAmplitudes(num_qubits=num_features, reps=1)
print("\nAnsatz Circuit:")
display(ansatz.draw('mpl'))

# Combine them to see the full circuit structure (optional, for visualization)
combined_circuit = feature_map.compose(ansatz)
print("\nCombined Feature Map and Ansatz:")
display(combined_circuit.draw('mpl'))
```

### 3. Configure Quantum Instance and Optimizer

We'll use a `BasicAer.get_backend('qasm_simulator')` for simulation and `COBYLA` as the classical optimizer.

```python
# Set random seed for reproducibility
algorithm_globals.random_seed = 12345

# Choose a classical optimizer
optimizer = COBYLA(maxiter=100) # Maximum iterations for optimization

# Choose a quantum backend (simulator for this example)
quantum_instance = QuantumInstance(BasicAer.get_backend('qasm_simulator'),
                                   shots=1024,
                                   seed_simulator=algorithm_globals.random_seed,
                                   seed_transpiler=algorithm_globals.random_seed)
```

### 4. Create and Train the VQC Model

Initialize the `VQC` with the feature map, ansatz, optimizer, and quantum instance. Then, train it with the training data.

```python
vqc = VQC(
    feature_map=feature_map,
    ansatz=ansatz,
    optimizer=optimizer,
    quantum_instance=quantum_instance
)

print("Training VQC...")
vqc.fit(X_train, y_train)
print("VQC training complete!")
```

### 5. Evaluate the VQC Model

Finally, evaluate the trained VQC on the test set to see its performance.

```python
train_score = vqc.score(X_train, y_train)
test_score = vqc.score(X_test, y_test)

print(f"VQC Training Accuracy: {train_score:.4f}")
print(f"VQC Test Accuracy: {test_score:.4f}")
```

### 6. Visualize Decision Boundary (Optional)

This helps to understand how the VQC separates the classes in the feature space.

```python
# Create a meshgrid to plot the decision boundary
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 50),
                     np.linspace(y_min, y_max, 50))

# Predict on the meshgrid points
Z = vqc.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot the decision boundary
plt.contourf(xx, yy, Z, alpha=0.8, cmap=plt.cm.RdBu)
plt.scatter(X_train[y_train == 0, 0], X_train[y_train == 0, 1], label='Train Class 0', marker='o', edgecolors='k')
plt.scatter(X_train[y_train == 1, 0], X_train[y_train == 1, 1], label='Train Class 1', marker='x', edgecolors='k')
plt.scatter(X_test[y_test == 0, 0], X_test[y_test == 0, 1], label='Test Class 0', marker='o', facecolors='none', edgecolors='green', s=70)
plt.scatter(X_test[y_test == 1, 0], X_test[y_test == 1, 1], label='Test Class 1', marker='x', facecolors='none', edgecolors='purple', s=70)

plt.title('VQC Decision Boundary')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
